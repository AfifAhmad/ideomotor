```python
import random

def generate_non_stop_codon():
    """Menghasilkan satu kodon yang bukan merupakan stop codon (TAA, TAG, TGA)."""
    bases = ['A', 'C', 'G', 'T']
    stop_codons = {'TAA', 'TAG', 'TGA'}
    while True:
        codon = ''.join(random.choice(bases) for _ in range(3))
        if codon not in stop_codons:
            return codon

def generate_dna_no_stops(num_codons=64):
    """Menghasilkan daftar kodon tanpa stop codon."""
    return [generate_non_stop_codon() for _ in range(num_codons)]

# Menghasilkan 64 kodon (total 192 residu)
codons = generate_dna_no_stops(64)
dna_192 = ''.join(codons)

# Membagi kodon ke dalam 8 baris (8 kodon per baris)
codons_per_line = 8
for i in range(0, len(codons), codons_per_line):
    line_codons = codons[i:i+codons_per_line]
    print(' '.join(line_codons))

print(f"\nTotal panjang karakter: {len(dna_192)} residu (Bebas Stop Codon: TAA, TAG, TGA)")
```

```python
import random

def mutate_sequence(original_codons):
    """Mengubah 1 sampai 4 kodon secara acak dan memastikan tidak ada stop codon baru."""
    bases = ['A', 'C', 'G', 'T']
    stop_codons = {'TAA', 'TAG', 'TGA'}

    # Salin list kodon asli
    mutated_codons = list(original_codons)

    # Tentukan jumlah kodon yang akan diubah (1 sampai 4)
    num_mutations = random.randint(1, 4)

    # Pilih indeks acak unik untuk dimutasi
    indices_to_mutate = random.sample(range(len(mutated_codons)), num_mutations)

    print(f"Melakukan mutasi pada {num_mutations} kodon di indeks: {indices_to_mutate}\n")

    for idx in indices_to_mutate:
        old_codon = mutated_codons[idx]
        while True:
            # Generate kodon baru
            new_codon = ''.join(random.choice(bases) for _ in range(3))
            # Pastikan bukan stop codon dan berbeda dari kodon lama
            if new_codon not in stop_codons and new_codon != old_codon:
                mutated_codons[idx] = new_codon
                print(f"Indeks {idx}: {old_codon} -> {new_codon}")
                break

    return mutated_codons

# Input sekuens DNA asal
dna_input = '''CTC TCT TAC TCT GCC TCT GTA GAG
AGA ACG AAT CAG AGC TTG AGG TGT
AAG TTC GCT CGG AGG TCT ACA ACA
GTA GGC ATC ACC TCT TGT TGC CCA
AGG ACA AAA AGG ACT TTT CCC TGC
TTG GAG GGA TCC CAC TTC CCG CTG
TAT ACT CGT TGC TTA TCA GAT CAC
ACG CCC ACA TGC GAA CCT TAT TTC
'''

# Split input berdasarkan spasi/baris baru untuk mendapatkan list kodon (3-mer)
codons_list = [codon.strip().upper() for codon in dna_input.split() if codon.strip()]

# Lakukan mutasi
mutated_codons_list = mutate_sequence(codons_list)
new_dna = ''.join(mutated_codons_list)

print("\n--- Hasil Setelah Mutasi ---")
codons_per_line = 8
for i in range(0, len(mutated_codons_list), codons_per_line):
    line_codons = mutated_codons_list[i:i+codons_per_line]
    print(' '.join(line_codons))

print(f"\nTotal panjang karakter setelah mutasi: {len(new_dna)} residu")
```
